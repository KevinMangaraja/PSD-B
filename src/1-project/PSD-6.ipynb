Penjelasan Alur Kerja (Workflow) Interpolasi dan Clustering pada KNIME

Workflow ini dirancang untuk menangani data dengan menggunakan teknik interpolasi untuk mengisi nilai yang hilang, kemudian melakukan segmentasi data menggunakan algoritma K-Means sebelum dan sesudah penerapan PCA (Principal Component Analysis).

Berikut adalah rincian langkah-langkah dari setiap node yang digunakan dalam alur kerja tersebut:

1. Excel Reader
Node ini berfungsi sebagai pintu masuk data, yang bertugas membaca dan mengimpor dataset dari format Excel ke dalam lingkungan KNIME. Data ini merupakan data mentah yang akan diproses lebih lanjut.

2. Missing Value (Interpolasi)
Node ini berfungsi untuk menangani data yang hilang (missing values). Dengan menggunakan metode interpolasi, node ini mengisi celah pada data berdasarkan nilai-nilai yang ada di sekitarnya. Hal ini sangat penting untuk menjaga integritas data time series atau data berkelanjutan agar tetap konsisten tanpa harus membuang sampel data.

3. Table View (Pengecekan Data)
Node ini digunakan untuk menampilkan tabel data setelah proses interpolasi selesai. Langkah ini bertujuan untuk melakukan inspeksi visual guna memastikan bahwa tidak ada lagi nilai yang kosong dan data siap untuk tahap clustering.

4. k-Means (Skenario 1: Sebelum PCA)
Node ini menerapkan algoritma K-Means pada dataset asli yang telah diinterpolasi. Proses ini mengelompokkan data berdasarkan seluruh fitur yang tersedia untuk melihat struktur kelompok awal sebelum dilakukan reduksi dimensi.

5. PCA (Principal Component Analysis)
Node PCA digunakan untuk mereduksi dimensi data. Teknik ini mentransformasikan fitur-fitur yang banyak menjadi beberapa komponen utama yang paling mewakili variansi data. PCA membantu mengurangi noise dan mempermudah visualisasi serta analisis pada data yang kompleks.

6. k-Means (Skenario 2: Sesudah PCA)
Node ini menjalankan kembali algoritma K-Means, namun kali ini pada data yang telah diproses oleh PCA. Tujuannya adalah untuk mendapatkan hasil pengelompokan yang lebih optimal dan bersih setelah fitur-fitur yang kurang relevan disaring.

Kesimpulan
Dengan menggabungkan teknik interpolasi dan PCA, kita dapat memastikan kualitas data yang lebih baik dan hasil clustering yang lebih akurat serta lebih mudah diinterpretasikan.
